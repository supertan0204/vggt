```python
import torch

# 创建一个形状为 B x S x P x 2C 的张量，假设 B=4, S=3, P=5, C=2
B, S, P, C = 4, 3, 5, 2
tensor = torch.randn(B, S, P, 2 * C)

# 打印原始张量的形状
print(f"Original tensor shape: {tensor.shape}")

# 取[:,:,0]的操作
result = tensor[:, :, 3:]

# 打印结果的形状
print(f"Result tensor shape after slicing [:,:,0]: {result.shape}")
```

```python
def fib(max):
    n, a, b = 0, 0, 1
    while n < max:
        yield b
        a, b = b, a + b
        n = n + 1
    return 'done'
f = fib(6)
```
