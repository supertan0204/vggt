```python
def get_chunk_indices(batch_size, num_chunks, chunk_id):
    chunk_size = batch_size // num_chunks
    remainder = batch_size % num_chunks

    if remainder == 0: 
        start = chunk_id * chunk_size
        end = start + chunk_size - 1 
    else:
        if chunk_id < remainder:
            start = chunk_id * (chunk_size + 1)
            end = start + chunk_size
        else:
            # start = remainder * (chunk_size + 1) + (chunk_id - remainder) * chunk_size
            start = remainder + chunk_id * chunk_size
            end = start + chunk_size - 1
    return start, end

get_chunk_indices(13,3, 3)
```
